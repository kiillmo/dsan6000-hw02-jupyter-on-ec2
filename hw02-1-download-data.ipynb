# HW2 Part 2.1: Downloading the Hourly Wikipedia Event Data

Download the hourly `.parquet`-format Wikipedia event data from `s3://dsan6000-wikipedia` into a local `data` subfolder using **Boto3**.

These files are ephemeral: `data/` is listed in `.gitignore`, so nothing downloaded here is pushed to GitHub.

```python
import os
import boto3

SRC_BUCKET = "dsan6000-wikipedia"
SRC_PREFIX = "hourly_parquet/"
DATA_DIR = "data"

os.makedirs(DATA_DIR, exist_ok=True)
s3 = boto3.client("s3")
```

## List the hourly parquet files in the source bucket

A paginator is used rather than a single `list_objects_v2` call, since that call returns at most 1000 keys per response.

```python
paginator = s3.get_paginator("list_objects_v2")
keys = []
for page in paginator.paginate(Bucket=SRC_BUCKET, Prefix=SRC_PREFIX):
    for obj in page.get("Contents", []):
        if obj["Key"].endswith(".parquet"):
            keys.append(obj["Key"])

keys = sorted(keys)
print(f"Found {len(keys)} parquet files")
keys[:5]
```

## Download each file into the local `data` subfolder

```python
for key in keys:
    filename = os.path.basename(key)
    dest = os.path.join(DATA_DIR, filename)
    s3.download_file(SRC_BUCKET, key, dest)
    print(f"downloaded {filename}")
```

## Verify the download

```python
files = sorted(f for f in os.listdir(DATA_DIR) if f.endswith(".parquet"))
total_mb = sum(os.path.getsize(os.path.join(DATA_DIR, f)) for f in files) / 1e6
print(f"{len(files)} files, {total_mb:.1f} MB total")
print(f"first: {files[0]}")
print(f"last:  {files[-1]}")
```
